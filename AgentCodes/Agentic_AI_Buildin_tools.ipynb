```python
from langchain_core.tools import tool

@tool
def multiply(a: int, b: int) -> int:
    """Multiplies two numbers."""
    return a * b
print("Multiplication Tool Details")
print(f"Tool Name: {multiply.name}")
print(f"Tool Description: {multiply.description}")
print(f"Tool Function Signature: {multiply.args}")

@tool
def get_employee_email(employee_name:str)->str:
    """Fetches the email address of an employee given their name."""
    #Mock Database
    db={"Venkatesh":"rcvenky26@gmail.com","Vishnu":"vishnu2494@gmail.com"}
    return db.get(employee_name,"Employee Not Found")
print("Employee Email Tool Details")
print(f"Tool Name: {get_employee_email.name}")
print(f"Tool Description: {get_employee_email.description}")
print(f"Tool Function Signature: {get_employee_email.args}")
```

```python
from langchain_community.tools.tavily_search import TavilySearchResults
from dotenv import load_dotenv
load_dotenv() # Load environment variables from .env file
import os
api_key = os.getenv("TAVILY_API_KEY")
os.environ["TAVILY_API_KEY"] = api_key


#Create Tavily Search Tool
tavily_search_tool = TavilySearchResults(max_results=5)

#Test the Tavily Search Tool
tavily_result1 = tavily_search_tool.invoke("What is the current stock price of Apple Inc. (AAPL)?")
print("Tavily Search Success")
print(tavily_result1[0]["content"])
```

```python
tavily_result2 = tavily_search_tool.invoke("What is Chief Minister of Tamilnadu in 2026?")
print("Tavily Search Success...")
print(tavily_result2[0]["content"])
```
